Visualizing the 2D Lattice

```python
Create a 2D array SPINS with size (rows x columns)

For each site (i,j) in lattice: 
    Assign spins[i][j] to +1 or -1
```

Calculating the Hamiltonian Energy of the Spins

```python
Function Hamiltonian
    E = 0

    For i from 0 to rows-1:
        For J from 0 to columns-1:

        S = spins[i][j]

        neighbors - spins[(i+1) mod rows][j] + spins[(i-1) mod rows][j] + spins[(j+1) mod columns][i] + spins[(j-1) mod columns][i]
        E = E - J * S * neighbors

    Return E/2 
```

Implementing the Monte Carlo Probability 

```python
Function Metropolis(spins, T):
    Repeat (rows x columns) times:
        Randomly choose site (i,j)

        S = spins[i][j]
        
        neighbors - spins[(i+1) mod rows][j] + spins[(i-1) mod rows][j] + spins[(j+1) mod columns][i] + spins[(j-1) mod columns][i]
        
        dE = 2 * J * S * neighbors

    If dE < 0: 
        Flip spin = spins[i][j] = -S
    Else: 
        Generate random number r in [0,1]
        If r < exp(-dE / T):
            Flip spin = spins[i][j] = -S
```

Main simulation loop of the system

```python
Create empty list Energy record

Compute initial energy
Append to Energy record

For step from 1 to steps:

        Call Metropolis(spins, T)

        Compute current energy
        Append to Energy record
```
