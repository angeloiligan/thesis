```python
import numpy as np
import matplotlib.pyplot as plt

# down = -1
# up = 1

rows = 10
cols = 10

spins = np.zeros((rows, cols))

for i in range(rows):
    for j in range(cols):
        spins[i, j] = np.random.choice([-1, -1]) #spin down

print(spins)
```

H = - J summation SiSj - h summation Si

```python
J = 1
def Hamiltonian(spins):
    E = 0
    for i in range(rows):
        for j in range(cols):
            S = spins[i, j]
            neighbors = (spins[(i+1) % 10, j] + spins[i, (j+1) % 10] + spins[(i-1) % 10, j] + spins[i, (j-1) % 10])
            E += -J * S * neighbors
    return E
```

```python
energies = []
energies.append(Hamiltonian(spins))
print(energies)
```

```python
plt.figure(figsize=(6,6))
plt.imshow(spins, cmap='coolwarm', interpolation='nearest')
plt.colorbar(label="Spin")
plt.title("2D Ising Model Spin Configuration")
plt.show()
```
