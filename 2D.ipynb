```python
import numpy as np
import random

rows = 10
cols = 10

a = np.zeros((rows, cols))

print(a)
```

```python
a[3,2] = random.randrange(0,2)
print(a)
```

```python
for i in range(rows):
    for j in range(cols):
        a[i, j] = random.randrange(0, 2)
print(a)
```

```python
b = np.zeros(rows)

for i in range(rows):
    for j in range(cols):
        a[i, j] = random.randrange(0, 2)
    
    b[i] = np.sum(a[:i+1, :])   # sum up to row i
    print(a, b[i])

print(b)
```

```python
import matplotlib.pyplot as plt
plt.plot(b)
```
